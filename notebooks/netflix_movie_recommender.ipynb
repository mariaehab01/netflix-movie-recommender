```python
import pandas as pd
```

# Load datasets

```python
movies = pd.read_csv('/content/Netflix_Dataset_Movie.csv')
movies.head()
```

```python
ratings = pd.read_csv('/content/Netflix_Dataset_Rating.csv')
ratings.head()
```

# Merge the movie and ratings datasets


```python
movie_ratings=pd.merge(movies,ratings)
movie_ratings.head()
```

# Create pivot table: User_ID as index, Movie Name as columns, Ratings as values


```python
movie_ratings=movie_ratings.pivot_table(index=['User_ID'],columns=['Name'],values='Rating')
movie_ratings.head()
```

# Check if a movie contains the word 'Wild'

```python
movies['Name'][movies['Name'].str.contains('Wild ')]
```

# Choose a specific movie to find similar ones


```python
movieName = movie_ratings['Character']
movieName
```

# Find movies similar to the chosen one


```python
similar_movies = movie_ratings.corrwith(movieName)
similar_movies = similar_movies.dropna()
```

# Display top similar movies

```python
df = pd.DataFrame(similar_movies)
df.head(10)
```

#View sorted similarities

```python
similar_movies.sort_values(ascending=False)
```
